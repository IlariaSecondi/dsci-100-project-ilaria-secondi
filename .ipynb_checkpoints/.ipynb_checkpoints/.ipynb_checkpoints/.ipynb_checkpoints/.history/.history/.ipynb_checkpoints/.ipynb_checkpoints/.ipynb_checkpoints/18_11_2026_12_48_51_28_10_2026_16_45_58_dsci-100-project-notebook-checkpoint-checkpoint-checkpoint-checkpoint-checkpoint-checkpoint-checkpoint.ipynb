#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
library(tidymodels)
set.seed(2020) #setting the seed so this is recreatable

wmm_tidy <- wmm_tidy |> mutate(first_game_at_home = as.factor(first_game_at_home))

#Making my split: 75% for training and 25% for testing
wmm_split <- initial_split(wmm_tidy, prop = 0.75, strata = first_game_at_home)
wmm_training <- training(wmm_split)
wmm_testing <- testing(wmm_split)

#Creating a recipe, we are trying to predicg whether the first game was plyed at home using the number of tourney wins
wmm_recipe <- recipe(first_game_at_home ~ tourney_wins, data = wmm_training) |> step_scale(all_predictors()) |> 
    step_center(all_predictors())

#I want to test k-values from 1-201 in increments of 10, ex: 1, 11, 21, 31..., 201
gridvals <- tibble(neighbors = seq(from = 1, to = 201, by = 10))

#Creating a model, using the tune() function to evaluate the k-value that results in the greatest accuracy
wmm_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |> set_engine("kknn") |> set_mode("classification")

#I'm doing 5-fold cross-validation
wmm_vfold <- vfold_cv(wmm_training, v = 5, strata = first_game_at_home)

#Putting it all into a workflow and collecting metrics to find the accuracy of the model for each k-value I'm testing
wmm_fit <- workflow() |> add_recipe(wmm_recipe) |> add_model(wmm_spec) |> tune_grid(resamples = wmm_vfold, grid = gridvals) |> 
    collect_metrics()

wmm_results <- wmm_fit |> filter(.metric == "accuracy")

#Pulling the number of neighbors that yields the highest accuracy

wmm_best_k <- wmm_results |> arrange(desc(mean)) |> head(1) |> pull(neighbors)
```

```r
set.seed(2020) #setting the seed so this is recreatable
options(repr.plot.width = 6, repr.plot.height = 4)

#Making a plot to visualize accuracy vs k

wmm_results_plot <- wmm_results |> ggplot(aes(x = neighbors, y = mean)) + geom_point() + geom_line() +
    labs(x = "Number of Neighbors (k-value)", y = "Accuracy Estimate", title = "Accuracy of Model Versus Number of Neighbors") +
    theme(text = element_text(size = 15))

wmm_results_plot
```

```r
#The plot shows that a k-value of around 175 yields the greatest accuracy.
#Setting k to the number of neighbors that yields the highest accuracy, an object attributed to wmm_best_k

wmm_spec_best_k <- nearest_neighbor(weight_func = "rectangular", neighbors = wmm_best_k) |> set_engine("kknn") |> 
    set_mode("classification")

wmm_fit_best_k <- workflow() |> add_recipe(wmm_recipe) |> add_model(wmm_spec_best_k) |> fit(data = wmm_training)

#Finding the accuracy of the final model

wmm_test_predictions <- predict(wmm_fit_best_k, wmm_testing) |> bind_cols(wmm_testing) |>
    metrics(truth = first_game_at_home, estimate = .pred_class) |> filter(.metric == "accuracy")

wmm_test_predictions
```
